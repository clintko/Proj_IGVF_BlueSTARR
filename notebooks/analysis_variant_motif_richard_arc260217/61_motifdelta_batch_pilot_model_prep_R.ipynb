## Arrange table

```r
### clean the motif name
dat = dat_motifdelta_event
dat = dat %>% dplyr::mutate(
    Motif_Name = sub("^(\\S+) \\1$", "\\1", Motif_Name)
)

### assign and show
dat_motifdelta_event_arrange = dat
print(dim(dat))
fun_display_table(head(dat, 3))
```

```r
### Motif gain events
dat = dat_motifdelta_event_arrange 
dat = dat %>%
    dplyr::filter(Event_Type == "Gain") %>%
    dplyr::select(Variant_ID, Motif_Name, Delta_Gain = Delta) %>%
    dplyr::distinct()

### sanity check: each variant-motif pair already has one row
tmp = dat %>%
    dplyr::group_by(Variant_ID, Motif_Name) %>%
    dplyr::summarize(Count = n(), .groups = "drop")
vec = unique(tmp$Count)
cat("Each variant-motif pair has one row only?", vec == 1, "\n")

### assign and show
dat_event_gain = dat
print(dim(dat))
fun_display_table(head(dat))
```

```r
### Motif loss events
dat = dat_motifdelta_event_arrange 
dat = dat %>%
    dplyr::filter(Event_Type == "Loss") %>%
    dplyr::select(Variant_ID, Motif_Name, Delta_Loss = Delta) %>%
    dplyr::distinct()

### sanity check: each variant-motif pair already has one row
tmp = dat %>%
    dplyr::group_by(Variant_ID, Motif_Name) %>%
    dplyr::summarize(Count = n(), .groups = "drop")
vec = unique(tmp$Count)
cat("Each variant-motif pair has one row only?", vec == 1, "\n")

### assign and show
dat_event_loss = dat
print(dim(dat))
fun_display_table(head(dat))
```

```r
### Motif full events
dat = dplyr::full_join(
    dat_event_gain,
    dat_event_loss,
    by = c("Variant_ID", "Motif_Name")
    ) %>%
    dplyr::mutate(across(where(is.numeric), ~ replace_na(., 0))) %>%
    dplyr::mutate(Delta_Full = Delta_Gain + Delta_Loss)

### assign and show
dat_event_full = dat
print(dim(dat))
fun_display_table(head(dat))
```

```r
### summary of delta
dat = dat_event_full
dat = dat %>% 
    dplyr::summarize(
        across(
            where(is.numeric), 
            list(Min = min, Mean = mean, Median = median, Max = max),
            .names = "{.col}:{.fn}"
        ) 
    )
dat = dat %>%
    tidyr::pivot_longer(
        cols = everything(),               
        names_to  = c("Variable", ".value"), 
        names_sep = ":"
    )

### show column summary
fun_display_table(dat)
```

## Prepare X and Y for modeling

### y := Delta(BlueSTARR) vector

```r
dat = dat_region_variant
print(dim(dat))

dat = dat %>% 
    dplyr::select(Variant_ID, Delta) %>%
    dplyr::distinct()

dat_Y = dat
print(dim(dat))
head(dat)
```

### X := Delta(Motif) matrix for both gain and loss

```r
### Motif gain events
dat = dat_event_gain
dat = dat %>%
    tidyr::pivot_wider(
        id_cols     = Variant_ID,
        names_from  = Motif_Name,
        values_from = Delta_Gain,
        values_fill = 0   # NA -> 0
    )

### assign and show
dat_X_gain = dat
print(dim(dat))
fun_display_table(head(dat, 3))
```

```r
### Motif loss events
dat = dat_event_loss
dat = dat %>%
    tidyr::pivot_wider(
        id_cols     = Variant_ID,
        names_from  = Motif_Name,
        values_from = Delta_Loss,
        values_fill = 0   # NA -> 0
    )

### assign and show
dat_X_loss = dat
print(dim(dat))
fun_display_table(head(dat, 3))
```

```r
dat = dat_event_full
dat = dat %>%
    tidyr::pivot_wider(
        id_cols     = Variant_ID,
        names_from  = Motif_Name,
        values_from = Delta_Full,
        values_fill = 0   # NA -> 0
    )

### assign and show
dat_X_full = dat
print(dim(dat))
fun_display_table(head(dat))
```

### Join X and y

```r
### init: list all matrices
lst = list(
    "Full" = dat_X_full, 
    "Gain" = dat_X_gain,
    "Loss" = dat_X_loss
)

### join X, y
lst = lapply(lst, function(dat_X){
    dat = dplyr::inner_join(dat_Y, dat_X, by = "Variant_ID")
    return(dat)
})

### assign and show
lst_dat_XY = lst
print(lapply(lst, dim))
```

## Sanity check

**sanity check: remove all zero motifs or variants**

```r
### count zeros
lst = lst_dat_XY
for (idx in names(lst)){
    cat("Matrix:", idx, "\n")
    dat = lst[[idx]]
    dat = dat %>% dplyr::select(-Variant_ID, -Delta)
    
    vec = colSums(dat != 0)
    vec = names(vec)[vec == 0]
    cat("#{All zero motifs} =", length(vec), "\n")
    
    vec = rowSums(dat != 0)
    vec = names(vec)[vec == 0]
    cat("#{All zero variants} =", length(vec), "\n")
    cat("\n")
}
```

## Filter: drop “very rare” motifs

```r
lst = lst_dat_XY
lst = lapply(lst, function(dat){
    ### set threshold: at least 0.5% of variants
    num_feature   = nrow(dat)
    num_threshold = 0.005 * num_feature
    
    ### count non-zeros for each motif
    tmp = dat %>% dplyr::select(-Variant_ID, -Delta)
    mat = as.matrix(tmp)
    vec = colSums(mat != 0)
    
    ### filter motifs less than the threshold
    idx = names(vec)[vec >= num_threshold]
    mat = mat[, idx, drop = FALSE]

    ### Rebuild pruned dat_XY
    tmp = dat %>% select(Variant_ID, Delta)
    dat = bind_cols(tmp, mat)
    return(dat)
})

### assign and show
lst_dat_XY_filter = lst
cat("Before filter:\n")
print(lapply(lst_dat_XY, dim))
cat("After filter:\n")
print(lapply(lst_dat_XY_filter, dim))
```
