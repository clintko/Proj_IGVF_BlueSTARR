**Check**

```r
stopifnot(all(rownames(mat_event_ori_import) %in% dat_variant_bluestarr_import$Variant_UnObs_vs_Obs))
```

```r
fun_build_y = function(mat_X, dat_variant, col_id = "Variant_UnObs_vs_Obs", col_y = "Delta") {

    ### y table
    dat_y = dat_variant %>%
        dplyr::select(
            Variant_ID = dplyr::all_of(col_id),
            Delta_BlueSTARR = dplyr::all_of(col_y)
        ) %>%
        dplyr::distinct(Variant_ID, .keep_all = TRUE)

    ### align to X rownames
    vec_variant = rownames(mat_X)

    dat_y = dat_y %>%
        dplyr::filter(Variant_ID %in% vec_variant) %>%
        dplyr::mutate(Variant_ID = factor(Variant_ID, levels = vec_variant)) %>%
        dplyr::arrange(Variant_ID)

    ### extract numeric vector
    vec_y = dat_y$Delta_BlueSTARR
    names(vec_y) = as.character(dat_y$Variant_ID)

    ### sanity
    stopifnot(length(vec_y) == length(vec_variant))
    stopifnot(identical(names(vec_y), vec_variant))

    return(vec_y)
}

### --------------------------------
### 3) build y for X_ori and X_nuc
### --------------------------------
vec_y_ori = fun_build_y(mat_event_ori_import, dat_variant_bluestarr_import)

# vec_y_nuc should be identical because rownames are identical between X_ori and X_nuc
# but build it anyway for safety if you want:
vec_y_nuc = fun_build_y(mat_event_nuc_import, dat_variant_bluestarr_import)

###
print(all(vec_y_ori == vec_y_nuc))
cat("\n")
print(length(vec_y_ori))
cat("\n")
print(summary(vec_y_ori))
cat("\n")
print(head(vec_y_ori))
```
