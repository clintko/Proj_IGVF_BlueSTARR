**Import data**

```python
### set file directory
txt_fdiry = os.path.join(FD_DATA, "motif_jaspar2024")
txt_fname = "JASPAR2024_CORE_vertebrates_non-redundant.meme"
txt_fpath = os.path.join(txt_fdiry, txt_fname)

### import meme file
dct_motif_jaspar = mymeme.read_meme(txt_fpath)
```

**Summarize data**

```python
dct_motif = dct_motif_jaspar.copy()
first_key = next(iter(dct_motif))
first_pwm = dct_motif[first_key]

print("Type:",       type(dct_motif))
print("Num motifs:", len(dct_motif))
print("First motif key:",    first_key)
print("First PWM shape:",    first_pwm.shape)
print("First 3 rows:\n",     first_pwm[:3])
print("Row sums (first 3):", first_pwm[:3].sum(axis=1))
```

```python
arr_len = np.array([pwm.shape[0] for pwm in dct_motif_jaspar.values()])

print("Motif length statistics")
print("  count :", arr_len.size)
print("  min   :", arr_len.min())
print("  25%   :", np.percentile(arr_len, 25))
print("  median:", np.median(arr_len))
print("  mean  :", arr_len.mean())
print("  75%   :", np.percentile(arr_len, 75))
print("  max   :", arr_len.max())
```
