### Run script

expected output files:

````bash
### ------------------------------------------------------------
### set script & options
### ------------------------------------------------------------

### set script
FP_EXE="${FD_EXE}/run_xgboost_cv_train.sh"

### set slurm opt
NUM_CPU=11
NUM_MEM=30G
LST_OPTS=(
    -A "${SLURM_ACCOUNT}"
    -p "${SLURM_PARTITION}"
    --cpus-per-task="${NUM_CPU}"
    --mem="${NUM_MEM}"
    --chdir="${FD_EXE}"
    --export=ALL,FP_CNF="${FP_CNF}"
    --parsable
)

if [[ -n "${EXCLUDE_NODES}" ]]; then
    LST_OPTS+=( --exclude="${EXCLUDE_NODES}" )
fi

### ------------------------------------------------------------
### initialization
### ------------------------------------------------------------

### set parameters
FN_PREFIX="variant_closed_gof_bluestarr.flankL35R70.unobs_vs_obs"
FD_PREFIX="motifmodel_pilot"
TXT_MOTIF_SET="jvierstra_v2.1beta"
TXT_MOTIF_LAB="jvierstra"

TXT_FOLDER="${FD_PREFIX}_${TXT_MOTIF_SET}"
FD_INP="${FD_RES}/analysis_variant_motif_richard/${TXT_FOLDER}"
FD_OUT="${FD_INP}"
FP_INP_PARAMS="${FD_INP}/xbg_cv_train_params.json"

### show parameters
ls "${FD_INP}/xbg_cv_train_params.json"
echo 
cat "${FD_INP}/xbg_cv_train_params.json"
echo

### ------------------------------------------------------------
### submit jobs
### ------------------------------------------------------------

### Loop init
TXT_EXE="xgboost"
LST_SRC=(ori dinuc)
LST_JOBS=()

### Loop through I/O
for TXT_SRC in "${LST_SRC[@]}"; do
        
    ### set input/output
    FP_INP_PREFIX="${FD_INP}/${FN_PREFIX}.pilot"
    FP_OUT_PREFIX="${FD_OUT}/${FN_PREFIX}.pilot.${TXT_SRC}"

    FP_INP_Y="${FP_INP_PREFIX}.y.rds"
    FP_INP_X="${FP_INP_PREFIX}.${TXT_SRC}.X.rds"

    ### set job/log
    TXT_TAG="${TXT_SRC}"
    TXT_JOB="${TXT_EXE}.${TXT_MOTIF_LAB}.${TXT_FOLDER}.${TXT_TAG}"
    FN_LOG="run.${TXT_JOB}.txt"
    FP_LOG="${FD_LOG}/${FN_LOG}"
    
    ### execute
    JOBID=$(sbatch \
        "${LST_OPTS[@]}" \
        --job-name="${TXT_JOB}" \
        --output="${FP_LOG}" \
        "${FP_EXE}" \
            --fp_inp_x "${FP_INP_X}" \
            --fp_inp_y "${FP_INP_Y}" \
            --fp_inp_params "${FP_INP_PARAMS}" \
            --fp_out_prefix "${FP_OUT_PREFIX}" \
        | awk '{print $NF}'
    )
    echo "Submitted ${TXT_TAG}: ${JOBID}"
    echo "LOG: \${FD_LOG}/${FN_LOG}"
    LST_JOBS+=("${JOBID}")
    echo
done
````

Model training and evaluation

Gradient boosted regression models were trained using XGBoost to predict regulatory activity scores from motif-derived feature matrices. Model training was performed using five-fold cross-validation with a maximum of 5000 boosting rounds and early stopping enabled with a patience of 20 rounds. The objective function was squared error regression with root mean squared error (RMSE) used as the evaluation metric. The model used a maximum tree depth of 6, a learning rate (eta) of 0.1, subsampling of 0.8, and column sampling of 0.8. The histogram-based tree construction algorithm was used for computational efficiency.

For each model, the optimal number of boosting iterations was determined using cross-validation by selecting the iteration with the minimum mean test RMSE. For the model trained on the original motif feature matrix, early stopping occurred after 1845 iterations, with the optimal model obtained at iteration 1825 and a cross-validation RMSE of 0.0555. In contrast, the model trained on the dinucleotide-shuffled control matrix converged substantially earlier, with early stopping at 364 iterations and an optimal iteration of 344, achieving a cross-validation RMSE of 0.0717.

Final models were retrained on the full dataset using the optimal number of boosting iterations determined from cross-validation. Training performance metrics were also recorded for reference. The model trained on the original motif features achieved a training RMSE of 0.0366, mean absolute error (MAE) of 0.0261, and coefficient of determination (R²) of 0.772. In comparison, the model trained on dinucleotide-shuffled features achieved a training RMSE of 0.0643, MAE of 0.0455, and R² of 0.298.
