## Sanity check: SHAP and yhat

````r
## 1) identify bias column (usually last, named "BIAS")
stopifnot("BIAS" %in% colnames(mat_shap_ori_import))
idx_bias <- which(colnames(mat_shap_ori_import) == "BIAS")

## 2) reconstruct predictions from SHAP
vec_bias <- mat_shap_ori_import[, idx_bias]
mat_phi  <- mat_shap_ori_import[, -idx_bias, drop = FALSE]
vec_yhat_recon <- vec_bias + rowSums(mat_phi)

## 3) compare to saved yhat
stopifnot(length(vec_yhat_ori_import) == length(vec_yhat_recon))

diff <- vec_yhat_ori_import - vec_yhat_recon

cat("max_abs_diff =", max(abs(diff)), "\n")
cat("mean_abs_diff =", mean(abs(diff)), "\n")
cat("sd_diff =", sd(diff), "\n")

## optional: inspect a few cases
head(cbind(yhat_saved = vec_yhat_ori_import,
           yhat_recon = vec_yhat_recon,
           diff = diff))
````

## Step 1: Global motif importance

````r
### Step 1: Global motif importance
### Goal: Which motifs most influence ΔBlueSTARR predictions?
### Metric: mean(|SHAP|)
### Intepretation:
###   mean(|SHAP|) = global importance
###   Large values mean the motif strongly affects predictions.

### remove bias column
mat_shap = mat_shap_ori_import
idx_bias <- which(colnames(mat_shap) == "BIAS")
mat_phi  <- mat_shap[, -idx_bias]

### compute importance
df_imp <- data.frame(
    motif = colnames(mat_phi),
    mean_abs_shap = colMeans(abs(mat_phi)),
    mean_shap = colMeans(mat_phi)
)

### sort
df_imp <- df_imp[order(-df_imp$mean_abs_shap), ]

head(df_imp, 20)
````

## Step 2: Directionality of motif effects

````r
### Step 2: Directionality of motif effects
### Goal: Does the motif tend to increase or decrease ΔBlueSTARR?
### Metric: mean(SHAP)
### Add interpretation:
###   mean_SHAP > 0   → activating motif
###   mean_SHAP < 0   → repressing motif

### sort by mean_SHAP > 0
df_imp <- df_imp[order(-df_imp$mean_shap), ]
fun_display_table(head(df_imp, 20))

### sort by mean_SHAP < 0
df_imp <- df_imp[order(df_imp$mean_shap), ]
fun_display_table(head(df_imp, 20))
````
