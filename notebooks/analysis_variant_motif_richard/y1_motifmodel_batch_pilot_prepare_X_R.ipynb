**Helper function**

````r
fun_parse_motif = function(dat, col = "Motif_Name") {
    dat = dat %>% dplyr::mutate(
        ### deduplicate exact "X X" case
        "{col}" := str_replace(.data[[col]], "^(\\S+)\\s+\\1$", "\\1"),
    
        ### Motif_Index: take everything before the first ':'
        Motif_Index = str_extract(.data[[col]], "^[^:]+"),
    
        ### Motif_Label: take everything after the first ':'
        Motif_Label = str_replace(.data[[col]], "^[^:]+:", "")
    )
    return(dat)
}
````

````r
lst = lst_dat_event_concat
lst = lapply(lst, fun_parse_motif)
lst = lapply(lst, function(dat){
    dat = dat %>%
        ### merge by strand
        dplyr::group_by(Variant_ID, Motif_Name, Motif_Index, Motif_Label, Event_Type) %>%
        dplyr::summarise(Delta = Delta[which.max(abs(Delta))], .groups = "drop") %>%
        ### merge by event type
        dplyr::group_by(Variant_ID, Motif_Name, Motif_Index, Motif_Label) %>%
        dplyr::summarise(Delta = Delta[which.max(abs(Delta))], .groups = "drop")
    return(dat)
})

### assign and show
lst_dat_event_arrange = lst

res = lapply(lst, dim)
print(res)

dat = lst[[1]]
print(dim(dat))
fun_display_table(head(dat))
````

````r
fun_build_X = function(dat_long, vec_variant, vec_motif) {

    ### load as data.table and subset
    dat = data.table::as.data.table(dat_long)
    dat = dat[
        Variant_ID %in% vec_variant &
        Motif_Name %in% vec_motif,
        .(Variant_ID, Motif_Name, Delta)
    ]

    ### cast to wide
    ### note: dcast only creates columns that appear
    dat_wide = data.table::dcast(
        dat,
        Variant_ID ~ Motif_Name,
        value.var = "Delta",
        fill = 0
    )

    ### ensure all requested variants exist (add missing as all-zero rows)
    tmp = data.table::data.table(Variant_ID = vec_variant)
    dat_wide = merge(
        tmp,
        dat_wide,
        by = "Variant_ID",
        all.x = TRUE,
        sort = FALSE
    )

    ### fill NA with 0 (variants with no events)
    vec_col = setdiff(names(dat_wide), "Variant_ID")
    for (txt_col in vec_col) {
        idx = which(is.na(dat_wide[[txt_col]]))
        if (length(idx) > 0) {
            data.table::set(dat_wide, idx, txt_col, 0)
        }
    }

    ### ensure all requested motifs exist as columns (add missing as 0)
    vec_missing = setdiff(vec_motif, names(dat_wide))
    if (length(vec_missing) > 0) {
        for (txt_motif in vec_missing) {
            dat_wide[, (txt_motif) := 0]
        }
    }

    ### reorder rows to match vec_variant
    data.table::setkey(dat_wide, Variant_ID)
    dat_wide = dat_wide[data.table::data.table(Variant_ID = vec_variant)]

    ### reorder columns to match vec_motif
    ### keep as numeric matrix
    mat = as.matrix(dat_wide[, ..vec_motif])
    rownames(mat) = dat_wide$Variant_ID

    ### sanity check
    stopifnot(nrow(mat) == length(vec_variant))
    stopifnot(ncol(mat) == length(vec_motif))

    return(mat)
}
````

````r
vec_txt_motif = sort(Reduce(union, lapply(lst_dat_event_arrange, \(d) unique(d$Motif_Name))))
length(vec_txt_motif)  # should be 629
````

````r
vec_variants_top <- sort(unique(lst_dat_event_arrange$Top_Ori$Variant_ID))
vec_variants_low <- sort(unique(lst_dat_event_arrange$Low_Ori$Variant_ID))

length(vec_variants_top)  # should be 100000
length(vec_variants_low)  # should be 100000
````

````r
vec_variants <- c(vec_variants_top, vec_variants_low)  # 200k
vec_motifs   <- vec_txt_motif                           # 629

X_top_ori <- fun_build_X(lst_dat_event_arrange$Top_Ori, vec_variants_top, vec_motifs)
X_low_ori <- fun_build_X(lst_dat_event_arrange$Low_Ori, vec_variants_low, vec_motifs)

X_top_nuc <- fun_build_X(lst_dat_event_arrange$Top_Nuc, vec_variants_top, vec_motifs)
X_low_nuc <- fun_build_X(lst_dat_event_arrange$Low_Nuc, vec_variants_low, vec_motifs)

X_ori <- rbind(X_top_ori, X_low_ori)
X_nuc <- rbind(X_top_nuc, X_low_nuc)
````

````r
# sanity: same dimensions and same columns
stopifnot(identical(colnames(X_ori), colnames(X_nuc)))

# 200k × 629 expected
print(dim(X_ori))
print(dim(X_nuc))

# check those 5 motifs exist as columns
vec = c("AC0500:ZNF:C2H2_ZF","AC0344:ZNF:C2H2_ZF","AC0235:HINFP:C2H2_ZF","AC0346:ZNF:C2H2_ZF","AC0405:ZNF:C2H2_ZF")
print(vec %in% colnames(X_ori))

# check missing motifs are indeed zero-filled where absent
print(colSums(X_ori[, vec, drop=FALSE] != 0))
````
