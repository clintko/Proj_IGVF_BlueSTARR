```r
fun_parse_motif = function(dat, col = "Motif_Name") {
    dat = dat %>% dplyr::mutate(
        ### deduplicate exact "X X" case
        "{col}" := str_replace(.data[[col]], "^(\\S+)\\s+\\1$", "\\1"),
    
        ### Motif_Index: take everything before the first ':'
        Motif_Index = str_extract(.data[[col]], "^[^:]+"),
    
        ### Motif_Label: take everything after the first ':'
        Motif_Label = str_replace(.data[[col]], "^[^:]+:", "")
    )
    return(dat)
}
```

```r
dat_event_full      = dat_motifdelta_top_event_arrange
dat_summary_motif   = dat_motifdelta_top_motif_arrange
dat_summary_variant = dat_motifdelta_top_variant_import
```

```r
### Choose which variants/motifs to plot
n_var   = 200
n_motif = 100

### pick variants: most events
vec_txt_variant = dat_summary_variant %>%
    dplyr::mutate(Total = Count_Gain + Count_Loss) %>%
    dplyr::arrange(desc(Total)) %>%
    dplyr::slice_head(n = n_var) %>%
    dplyr::pull(Variant_ID)

### pick motifs: most gain/loss (choose one criterion or Total)
vec_txt_motif = dat_summary_motif %>%
    dplyr::mutate(Total = Count_Gain + Count_Loss) %>%
    dplyr::arrange(desc(Total)) %>%
    dplyr::slice_head(n = n_motif) %>%
    dplyr::pull(Motif_Name)

### filter events to subset
dat_event_sub = dat_event_full %>%
    dplyr::filter(
        Variant_ID %in% vec_txt_variant, 
        Motif_Name %in% vec_txt_motif
    )

cut_prob = 0.7
cut_gain = quantile(dat_event_sub$Delta[dat_event_sub$Event_Type == "Gain"], probs = cut_prob, na.rm = TRUE)
cut_loss = quantile(dat_event_sub$Delta[dat_event_sub$Event_Type == "Loss"], probs = 1 - cut_prob, na.rm = TRUE)

### Build gain and loss matrices (wide format)
mat_event_gain = dat_event_sub %>%
    dplyr::filter(Event_Type == "Gain", Delta >= cut_gain) %>%
    dplyr::select(Variant_ID, Motif_Name, Delta) %>%
    tidyr::pivot_wider(names_from = Motif_Name, values_from = Delta, values_fill = NA_real_) %>%
    column_to_rownames("Variant_ID") %>%
    as.matrix()

mat_event_loss <- dat_event_sub %>%
    dplyr::filter(Event_Type == "Loss", Delta <= cut_loss) %>%
    #dplyr::mutate(DeltaAbs = abs(Delta)) %>%
    #dplyr::select(Variant_ID, Motif_Name, DeltaAbs) %>%
    #tidyr::pivot_wider(names_from = Motif_Name, values_from = DeltaAbs, values_fill = NA_real_) %>%
    dplyr::select(Variant_ID, Motif_Name, Delta) %>%
    tidyr::pivot_wider(names_from = Motif_Name, values_from = Delta, values_fill = NA_real_) %>%
    column_to_rownames("Variant_ID") %>%
    as.matrix()
```

```r
drop_sparse <- function(mat, min_non_na = 2) {
  mat <- mat[rowSums(!is.na(mat)) >= min_non_na, , drop = FALSE]
  mat <- mat[, colSums(!is.na(mat)) >= min_non_na, drop = FALSE]
  return(mat)
}

mat_event_gain2 = drop_sparse(mat_event_gain, min_non_na = 2)
mat_event_loss2 = drop_sparse(mat_event_loss, min_non_na = 2)

mat_gain_plot = mat_event_gain2
mat_gain_plot[is.na(mat_gain_plot)] = 0

mat_loss_plot = mat_event_loss2
mat_loss_plot[is.na(mat_loss_plot)] = 0
```

```r
vec_txt_color = colorRampPalette(brewer.pal(9, "Reds"))(100)

plt = pheatmap(
    mat_gain_plot,
    color = vec_txt_color,
    border_color = NA,
    clustering_distance_rows = "correlation",
    clustering_distance_cols = "correlation",
    cluster_rows = TRUE,
    cluster_cols = TRUE,
    show_rownames = FALSE,
    #fontsize_row = 6,
    fontsize_col = 8,
    main = sprintf("Strong Gain Δ (>= %.3f; q=%.2f)", cut_gain, cut_prob)
)

options(repr.plot.height=10, repr.plot.width=12)
print(plt)
```

```r
vec_txt_color = colorRampPalette(brewer.pal(9, "Blues"))(100)
vec_txt_color = rev(vec_txt_color)

plt = pheatmap(
    mat_loss_plot,
    color = vec_txt_color,
    border_color = NA,
    clustering_distance_rows = "correlation",
    clustering_distance_cols = "correlation",
    cluster_rows = TRUE,
    cluster_cols = TRUE,
    show_rownames = FALSE,
    #fontsize_row = 6,
    fontsize_col = 8,
    main = sprintf("Strong Loss Δ (<= %.3f; q=%.2f)", cut_loss, cut_prob)
)

options(repr.plot.height=10, repr.plot.width=10)
print(plt)
```

## Explore: heatmap

```r
fun_heatmap_table = function(
    dat_event,
    dat_motif,
    dat_variant,
    event_type = c("Gain", "Loss"),
    n_var   = 50,
    n_motif = 50,
    motif_name_policy = c("top_total", "first")  # how to pick a representative name/index per label
){
    event_type = match.arg(event_type)
    motif_name_policy = match.arg(motif_name_policy)

    # ----------------------------
    # 1) pick top variants
    # ----------------------------
    vec_txt_variant = dat_variant %>%
        dplyr::arrange(dplyr::desc(if (event_type == "Gain") Count_Gain else Count_Loss)) %>%
        dplyr::slice_head(n = n_var) %>%
        dplyr::pull(Variant_ID)

    # ----------------------------
    # 2) pick top motif labels (IMPORTANT: use unique labels)
    # ----------------------------
    # dat_motif should have Motif_Label, Motif_Name, Motif_Index, Count_Gain, Count_Loss
    dat_label_summary = dat_motif %>%
        dplyr::group_by(Motif_Label) %>%
        dplyr::summarise(
            Count_Gain = sum(Count_Gain, na.rm = TRUE),
            Count_Loss = sum(Count_Loss, na.rm = TRUE),
            .groups = "drop"
        )

    vec_txt_motif_label = dat_label_summary %>%
        dplyr::arrange(dplyr::desc(if (event_type == "Gain") Count_Gain else Count_Loss)) %>%
        dplyr::slice_head(n = n_motif) %>%
        dplyr::pull(Motif_Label)

    # ----------------------------
    # 3) count events per (Variant_ID, Motif_Label)
    # ----------------------------
    dat_event_sub = dat_event %>%
        dplyr::filter(
            Event_Type == event_type,
            Variant_ID  %in% vec_txt_variant,
            Motif_Label %in% vec_txt_motif_label
        ) %>%
        dplyr::count(Variant_ID, Motif_Label, name = "Count")

    # complete missing cells
    dat_event_sub = dat_event_sub %>%
        tidyr::complete(
            Variant_ID  = vec_txt_variant,
            Motif_Label = vec_txt_motif_label,
            fill = list(Count = 0)
        )

    # ----------------------------
    # 4) build a label -> (Motif_Index, Motif_Name) mapping
    #    (one representative per label)
    # ----------------------------
    if (motif_name_policy == "top_total") {
        dat_label_map = dat_event %>%
            dplyr::filter(Event_Type == event_type, Motif_Label %in% vec_txt_motif_label) %>%
            dplyr::count(Motif_Label, Motif_Index, Motif_Name, name = "TotalEvents") %>%
            dplyr::arrange(Motif_Label, dplyr::desc(TotalEvents)) %>%
            dplyr::group_by(Motif_Label) %>%
            dplyr::slice_head(n = 1) %>%
            dplyr::ungroup() %>%
            dplyr::select(Motif_Label, Motif_Index, Motif_Name)
    } else {
        dat_label_map = dat_event %>%
            dplyr::filter(Motif_Label %in% vec_txt_motif_label) %>%
            dplyr::distinct(Motif_Label, Motif_Index, Motif_Name) %>%
            dplyr::group_by(Motif_Label) %>%
            dplyr::slice_head(n = 1) %>%
            dplyr::ungroup()
    }

    # attach metadata (repeated per cell)
    dat_event_sub = dat_event_sub %>%
        dplyr::left_join(dat_label_map, by = "Motif_Label")

    # ----------------------------
    # 5) order factors
    # ----------------------------
    ord_variant = dat_event_sub %>%
        dplyr::group_by(Variant_ID) %>%
        dplyr::summarise(Total = sum(Count), .groups = "drop") %>%
        dplyr::arrange(dplyr::desc(Total)) %>%
        dplyr::pull(Variant_ID)

    ord_motif = dat_event_sub %>%
        dplyr::group_by(Motif_Label) %>%
        dplyr::summarise(Total = sum(Count), .groups = "drop") %>%
        dplyr::arrange(dplyr::desc(Total)) %>%
        dplyr::pull(Motif_Label)

    dat_event_sub = dat_event_sub %>%
        dplyr::mutate(
            Variant_ID  = factor(Variant_ID,  levels = ord_variant),
            Motif_Label = factor(Motif_Label, levels = ord_motif)
        )

    return(dat_event_sub)
}
```

```r
fun_plot_heatmap = function(dat_plot, title, vec_txt_color) {
    
    gpt = ggplot(dat_plot, aes(x = Motif_Label, y = Variant_ID, fill = Count)) +
        geom_tile() +
        scale_fill_gradientn(
            colours = vec_txt_color,
            name = "Event\nCount"
        ) +
        labs(title = title, x = "Motif label", y = "Variant") +
        theme_minimal(base_size = 12) +
        theme(
            axis.text.x = element_text(angle = 60, hjust = 1, vjust = 1),
            panel.grid  = element_blank()
        )
    return(gpt)
}
```

```r
### init
N_VAR   = 10
N_MOTIF = 10

### set table
dat_event   = dat_motifdelta_top_event_arrange
dat_motif   = dat_motifdelta_top_motif_arrange
dat_variant = dat_motifdelta_top_variant_import

### generate heatmap
vec_txt_color = colorRampPalette(brewer.pal(9, "Reds"))(100)
dat_gain = fun_heatmap_table(
    dat_event, dat_motif, dat_variant, 
    "Gain", 
    n_var   = N_VAR, 
    n_motif = N_MOTIF)

dat = dat_gain %>% dplyr::arrange(-Count, )
print(dim(dat))
dat
```

```r
# ----------------------------
# 3) Pick top variants/motifs (gain)
# ----------------------------
n_var   <- 50
n_motif <- 50

vec_variant_gain <- dat_motifdelta_summary_variant %>%
  arrange(desc(Count_Gain)) %>%
  slice_head(n = n_var) %>%
  pull(Variant_ID)

# You can choose Motif_Name or Motif_Label.
# If you use Motif_Label you might merge multiple motifs; for now use Motif_Name.
vec_motif_gain <- dat_motifdelta_summary_motif %>%
  arrange(desc(Count_Gain)) %>%
  slice_head(n = n_motif) %>%
  pull(Motif_Name)

# ----------------------------
# 4) Filter to gain events in subset
# ----------------------------
dat_gain_sub <- dat_motifdelta_event_full %>%
  filter(
    Event_Type == "Gain",
    Variant_ID %in% vec_variant_gain,
    Motif_Name %in% vec_motif_gain
  )

# ----------------------------
# 5) Collapse duplicates per (Variant, Motif)
#    (there can be multiple positions/strands)
#    For a heatmap, keep the MAX Delta per pair
# ----------------------------
dat_gain_sub2 <- dat_gain_sub %>%
  group_by(Variant_ID, Motif_Name) %>%
  summarise(
    Delta = max(Delta, na.rm = TRUE),
    Motif_Label = first(Motif_Label),
    .groups = "drop"
  )

# ----------------------------
# 6) Wide matrix: rows=Variant, cols=Motif
# ----------------------------
mat_gain <- dat_gain_sub2 %>%
  select(Variant_ID, Motif_Name, Delta) %>%
  pivot_wider(
    names_from  = Motif_Name,
    values_from = Delta,
    values_fill = NA_real_
  )

mat_gain <- as.data.frame(mat_gain)
rownames(mat_gain) <- mat_gain$Variant_ID
mat_gain$Variant_ID <- NULL
mat_gain <- as.matrix(mat_gain)

# Optional: simplify column names to Motif_Label for plotting
# (but keep them unique, because multiple Motif_Name can share a label)
colnames(mat_gain) <- str_replace(colnames(mat_gain), ":([^:]+)$", "")  # same as Motif_Label
colnames(mat_gain) <- make.unique(colnames(mat_gain), sep = "_")


# 1) Drop rows/cols that are entirely NA
keep_rows <- rowSums(!is.na(mat_gain)) > 0
keep_cols <- colSums(!is.na(mat_gain)) > 0
mat_gain2 <- mat_gain[keep_rows, keep_cols, drop = FALSE]

# (optional) also drop rows/cols with zero variance (can also break correlation distances)
keep_rows2 <- apply(mat_gain2, 1, function(x) sd(x, na.rm = TRUE) > 0)
keep_cols2 <- apply(mat_gain2, 2, function(x) sd(x, na.rm = TRUE) > 0)
mat_gain2 <- mat_gain2[keep_rows2, keep_cols2, drop = FALSE]

# sanity check
dim(mat_gain2)
```
