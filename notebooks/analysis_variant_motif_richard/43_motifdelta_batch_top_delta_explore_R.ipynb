```r
### Choose which variants/motifs to plot
n_var   = 200
n_motif = 100

### pick variants: most events
vec_txt_variant = dat_summary_variant %>%
    dplyr::mutate(Total = Count_Gain + Count_Loss) %>%
    dplyr::arrange(desc(Total)) %>%
    dplyr::slice_head(n = n_var) %>%
    dplyr::pull(Variant_ID)

### pick motifs: most gain/loss (choose one criterion or Total)
vec_txt_motif = dat_summary_motif %>%
    dplyr::mutate(Total = Count_Gain + Count_Loss) %>%
    dplyr::arrange(desc(Total)) %>%
    dplyr::slice_head(n = n_motif) %>%
    dplyr::pull(Motif_Name)

### filter events to subset
dat_event_sub = dat_event_full %>%
    dplyr::filter(
        Variant_ID %in% vec_txt_variant, 
        Motif_Name %in% vec_txt_motif
    )

cut_prob = 0.7
cut_gain = quantile(dat_event_sub$Delta[dat_event_sub$Event_Type == "Gain"], probs = cut_prob, na.rm = TRUE)
cut_loss = quantile(dat_event_sub$Delta[dat_event_sub$Event_Type == "Loss"], probs = 1 - cut_prob, na.rm = TRUE)

### Build gain and loss matrices (wide format)
mat_event_gain = dat_event_sub %>%
    dplyr::filter(Event_Type == "Gain", Delta >= cut_gain) %>%
    dplyr::select(Variant_ID, Motif_Name, Delta) %>%
    tidyr::pivot_wider(names_from = Motif_Name, values_from = Delta, values_fill = NA_real_) %>%
    column_to_rownames("Variant_ID") %>%
    as.matrix()

mat_event_loss <- dat_event_sub %>%
    dplyr::filter(Event_Type == "Loss", Delta <= cut_loss) %>%
    #dplyr::mutate(DeltaAbs = abs(Delta)) %>%
    #dplyr::select(Variant_ID, Motif_Name, DeltaAbs) %>%
    #tidyr::pivot_wider(names_from = Motif_Name, values_from = DeltaAbs, values_fill = NA_real_) %>%
    dplyr::select(Variant_ID, Motif_Name, Delta) %>%
    tidyr::pivot_wider(names_from = Motif_Name, values_from = Delta, values_fill = NA_real_) %>%
    column_to_rownames("Variant_ID") %>%
    as.matrix()
```

```r
drop_sparse <- function(mat, min_non_na = 2) {
  mat <- mat[rowSums(!is.na(mat)) >= min_non_na, , drop = FALSE]
  mat <- mat[, colSums(!is.na(mat)) >= min_non_na, drop = FALSE]
  return(mat)
}

mat_event_gain2 = drop_sparse(mat_event_gain, min_non_na = 2)
mat_event_loss2 = drop_sparse(mat_event_loss, min_non_na = 2)

mat_gain_plot = mat_event_gain2
mat_gain_plot[is.na(mat_gain_plot)] = 0

mat_loss_plot = mat_event_loss2
mat_loss_plot[is.na(mat_loss_plot)] = 0
```

```r
make_event_matrix <- function(dat_event, event_type, motifs_keep, order_by_total = TRUE) {

  dat_sub <- dat_event %>%
    dplyr::filter(Event_Type == event_type) %>%
    dplyr::filter(Motif_Name %in% motifs_keep)

  # order variants by total Delta within this event type (optional)
  if (order_by_total) {
    variant_order <- dat_sub %>%
      dplyr::group_by(Variant_ID) %>%
      dplyr::summarize(Total_Delta = sum(Delta, na.rm = TRUE), .groups = "drop") %>%
      dplyr::arrange(dplyr::desc(Total_Delta)) %>%
      dplyr::pull(Variant_ID)
  } else {
    variant_order <- sort(unique(dat_event$Variant_ID))
  }

  mat_wide <- dat_sub %>%
    dplyr::mutate(
      Variant_ID = factor(Variant_ID, levels = variant_order),
      Motif_Name = factor(Motif_Name, levels = motifs_keep)
    ) %>%
    dplyr::group_by(Variant_ID, Motif_Name) %>%
    dplyr::summarize(
      Delta = Delta[which.max(Delta)],  # in case duplicates, keep max
      .groups = "drop"
    ) %>%
    tidyr::pivot_wider(
      names_from  = Motif_Name,
      values_from = Delta,
      values_fill = 0
    )

  mat <- mat_wide %>%
    dplyr::select(-Variant_ID) %>%
    as.matrix()

  rownames(mat) <- as.character(mat_wide$Variant_ID)

  list(mat = mat, variant_order = variant_order)
}
```

```r
res_gain <- make_event_matrix(
  dat_event = dat_event_import,
  event_type = "Gain",
  motifs_keep = top_motifs,
  order_by_total = TRUE
)

res_loss <- make_event_matrix(
  dat_event = dat_event_import,
  event_type = "Loss",
  motifs_keep = top_motifs,
  order_by_total = TRUE
)

mat_gain <- res_gain$mat
mat_loss <- res_loss$mat
```
