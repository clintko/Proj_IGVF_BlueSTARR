## Prepare

**Convert X to sparse**

```r
### convert to sparse dgCMatrix
mat_X = mat_event_ori_import
vec_y = vec_y_ori_import

mat_X_sparse = Matrix::Matrix(mat_X, sparse = TRUE)

stopifnot(nrow(mat_X_sparse) == length(vec_y))
rm(mat_event_ori_import)
gc()

print(class(mat_X_sparse))
print(dim(mat_X_sparse))
```

**Build DMatrix**

```r
dtrain = xgboost::xgb.DMatrix(
    data  = mat_X_sparse,
    label = vec_y
)
```
