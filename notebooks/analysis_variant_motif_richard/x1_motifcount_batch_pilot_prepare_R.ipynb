## Arrange table

**Helper function**

```r
fun_parse_motif = function(dat, col = "Motif_Name") {
    dat = dat %>% dplyr::mutate(
        ### deduplicate exact "X X" case
        "{col}" := str_replace(.data[[col]], "^(\\S+)\\s+\\1$", "\\1"),
    
        ### Motif_Index: take everything before the first ':'
        Motif_Index = str_extract(.data[[col]], "^[^:]+"),
    
        ### Motif_Label: take everything after the first ':'
        Motif_Label = str_replace(.data[[col]], "^[^:]+:", "")
    )
    return(dat)
}
```

```r
lst = list(
    "Top_Ori" = dat_motifdelta_top_ori_event_import,
    "Top_Nuc" = dat_motifdelta_top_nuc_event_import,
    "Low_Ori" = dat_motifdelta_low_ori_event_import,
    "Low_Nuc" = dat_motifdelta_low_nuc_event_import
)

lst = lapply(lst, fun_parse_motif)

lst_dat_motifdelta_event_arrange = lst
dat = lst[[1]]
fun_display_table(head(dat))
```

## Variant-level counts of motifs

**Get variant-level count table (Gain)**

```r
lst = lst_dat_motifdelta_event_arrange
lst = lapply(lst, function(dat){
    dat = dat %>%
        dplyr::filter(Event_Type == "Gain") %>%
        dplyr::group_by(Variant_ID, Motif_Name) %>%
        dplyr::summarise(Gain_Present = 1, .groups = "drop")
    return(dat)
})

lst_dat_event_gain = lst
dat = lst[[1]]
fun_display_table(head(dat))
```

**Get variant-level count table (Loss)**

```r
lst = lst_dat_motifdelta_event_arrange
lst = lapply(lst, function(dat){
    dat = dat %>%
        dplyr::filter(Event_Type == "Loss") %>%
        dplyr::group_by(Variant_ID, Motif_Name) %>%
        dplyr::summarise(Loss_Present = 1, .groups = "drop")
    return(dat)
})

lst_dat_event_loss = lst
dat = lst[[1]]
fun_display_table(head(dat))
```

**Total counts**

```r
lst = lst_dat_motifdelta_event_arrange
lst = lapply(lst, function(dat){
    vec = dat$Variant_ID
    vec = unique(vec)
    num = length(vec)
    return(num)
})

dat = data.frame(
    Group_Type  = names(lst),
    Count_Total = unlist(lst)
)

dat_event_count_total = dat
fun_display_table(dat)
```

**Sanity check**

```r
### check if there is any duplicate
### there shouldn't -> expect all FALSE
lst = lst_dat_event_gain
lst = lapply(lst, function(dat){
    res = any(duplicated(dat[, c("Variant_ID","Motif_Name")]))
    return(res)
})
print(lst)
```

```r
### check if there is any duplicate
### there shouldn't -> expect all FALSE
lst = lst_dat_event_loss
lst = lapply(lst, function(dat){
    res = any(duplicated(dat[, c("Variant_ID","Motif_Name")]))
    return(res)
})
print(lst)
```

## Summarize counts by motif

**count variants by motif**

```r
lst = lst_dat_event_merge
lst = lapply(lst, function(dat){
    dat = dat %>%
        dplyr::group_by(Group_Type, Motif_Name) %>%
        dplyr::summarise(
            Count = dplyr::n_distinct(Variant_ID), 
            .groups = "drop"
        )
    return(dat)
})

### assign and show
lst_dat_motif_count = lst
res = lapply(lst, dim)
print(res)
```

**Complete the count tables**

```r
lst = lst_dat_motif_count
lst = lapply(lst, function(dat_motif_count){
    
    ### get list of all motifs
    dat = dat_motif_count
    vec = unique(dat$Motif_Name)
    vec_txt_motif = vec

    ### get complete combination of conditions and motifs
    dat = tidyr::expand_grid(
        Group = c("Top","Low"),
        Type  = c("Ori","Nuc"),
        Motif_Name = vec_txt_motif
    )
    dat = dat %>% 
        dplyr::mutate(Group_Type = paste(Group, Type, sep = "_")) %>%
        dplyr::select(Group_Type, Group, Type, Motif_Name)
    dat_metainfo = dat

    ### complete count table
    dat = dat_metainfo
    
    tmp = dat_motif_count
    dat = dat %>%
        dplyr::left_join(tmp, by = c("Group_Type", "Motif_Name")) %>%
        dplyr::mutate(Count = ifelse(is.na(Count), 0, Count))
    
    tmp = dat_event_count_total
    tmp = tmp %>% dplyr::select(Group_Type, Count_Total)
    dat = dat %>% dplyr::left_join(tmp, by = "Group_Type")
    return(dat)
})

### assign and show
lst_dat_motif_count_complete = lst

res = lapply(lst, dim)
print(res)
```

```r
lst = lst_dat_motif_count_complete
dat = lst[["Count_Gain"]]
fun_display_table(head(dat))
```
