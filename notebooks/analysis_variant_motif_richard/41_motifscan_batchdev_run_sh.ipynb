```bash
### set script
FP_EXE=${FD_EXE}/run_motifscan_variant.py

### set input files
FD_FASTA=${FD_RES}/analysis_variant_motif_richard/batches_dev
FP_MOTIF=${FD_RES}/analysis_variant_motif_richard/JASPAR2024_CORE_vertebrates_non-redundant.pmap.npz
#FP_MOTIF=${FD_DATA}/jaspar2024/JASPAR2024_CORE_vertebrates_non-redundant.npz

### init
VEC_TXT_TYPE=("ref" "obs" "unobs")

### loop through each variant type
for TXT_TYPE in ${VEC_TXT_TYPE[@]}; do
    ### set input/output file
    PREFIX=variant_closed_gof_bluestarr_flank35_dev1000_${TXT_TYPE}
    FN_INP=${PREFIX}.fa
    FN_OUT=motifscan_${PREFIX}.pkl

    FP_INP=${FD_FASTA}/${FN_INP}
    FP_OUT=${FD_FASTA}/${FN_OUT}

    ### set log file
    FN_LOG=run_motifscan_${PREFIX}.txt
    FP_LOG=${FD_LOG}/${FN_LOG}
    
    ### set resource
    NUM_CPU=10
    NUM_MEM=20G
    NUM_CORE=10
    
    ### print input
    echo "Run ${TXT_TYPE}:"
    echo "  Input :" ${FN_INP}
    echo "  Output:" ${FN_OUT}
    echo "  Log   :" '${FD_LOG}'/${FN_LOG}
    echo

    ### submit job
    sbatch \
        -A majoroslab \
        -p igvf \
        --cpus-per-task=${NUM_CPU} \
        --mem=${NUM_MEM} \
        --output=${FP_LOG} \
        --chdir=${FD_EXE} \
        ${FP_APP} python ${FP_EXE} \
            --txt_fpath_fasta  ${FP_INP} \
            --txt_fpath_motif  ${FP_MOTIF} \
            --txt_fpath_output ${FP_OUT} \
            --num_core         ${NUM_CORE}
    echo
done
```
