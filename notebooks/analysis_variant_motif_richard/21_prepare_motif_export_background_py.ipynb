````python
from motifdelta.stats import background_zero_order
````

## Import data

````python
### import fasta seqeunce
txt_fdiry = os.path.join(FD_RES, "analysis_variant_motif_richard")
txt_fname = "variant_closed_gof_bluestarr.flankL35R70.unobs_vs_obs.ref.fa"
txt_fpath = os.path.join(txt_fdiry, txt_fname)

### read sequences into a list of strings
lst_rec_seq = [str(record.seq).upper() for record in SeqIO.parse(txt_fpath, "fasta")]

print(f"Loaded {len(lst_rec_seq)} sequences from {os.path.basename(txt_fpath)}")
print("Example sequence:", lst_rec_seq[0][:50], "...")
````

````python
bg = background_zero_order(lst_rec_seq, alphabet="ACGT")

print("Background nucleotide frequencies:")
for base, p in zip("ACGT", bg):
    print(f"  {base}: {p:.4f}")
````

## Export results

````python
### set directory
txt_fdiry = os.path.join(FD_RES, "analysis_variant_motif_richard")
txt_fname = "background_zero_order.tsv"
txt_fpath = os.path.join(txt_fdiry, txt_fname)

### export
dat = pd.DataFrame({"base": list("ACGT"), "freq": bg})
dat.to_csv(txt_fpath, sep="\t", index=False)
print(dat)
````

````python
### set directory
txt_fdiry = os.path.join(FD_RES, "analysis_variant_motif_richard")
txt_fname = "background_zero_order.npy"
txt_fpath = os.path.join(txt_fdiry, txt_fname)

### export
np.save(txt_fpath, bg)
print(bg)
````
