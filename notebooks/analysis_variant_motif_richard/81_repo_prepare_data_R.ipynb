## Import data

**Check file existence**

```r
### set file directory
txt_fdiry = file.path(FD_RES, "analysis_variant_motif_richard")
txt_fname = "variant_closed_gof_bluestarr*"
txt_fglob = file.path(txt_fdiry, txt_fname)

vec = Sys.glob(txt_fglob)
vec = basename(vec)
for (txt in vec){cat(txt, "\n")}
```

**Import table**

```r
### set file directory
txt_fdiry = file.path(FD_RES, "analysis_variant_motif_richard")
txt_fname = "variant_closed_gof_bluestarr.tsv"
txt_fpath = file.path(txt_fdiry, txt_fname)

### read table
dat = read_tsv(txt_fpath, show_col_types = FALSE)

### assign and show
dat_variant_bluestarr_import = dat
print(dim(dat))
fun_display_table(head(dat, 3))
```

## Arrange table

```r
dat = dat_variant_bluestarr_import
vec = c(
    "Chrom", "ChromStart", "ChromEnd", "Pos0", "Ref", "Obs", "Unobs", "Delta"
)
dat = dat %>% 
    dplyr::select(all_of(vec)) %>%
    dplyr::arrange(desc(Delta))

### assign and show
dat_variant_bluestarr_arrange = dat
print(dim(dat))
fun_display_table(head(dat))
```

## Get the top and bottom rows

```r
dat = dat_variant_bluestarr_arrange
num = 100000
dat_variant_bluestarr_top100k = dat %>% dplyr::slice_head(n = num)
dat_variant_bluestarr_low100k = dat %>% dplyr::slice_tail(n = num)
```

```r
dat = dat_variant_bluestarr_top100k
print(dim(dat))
fun_display_table(head(dat))
```

```r
dat = dat_variant_bluestarr_low100k
print(dim(dat))
fun_display_table(head(dat))
```
