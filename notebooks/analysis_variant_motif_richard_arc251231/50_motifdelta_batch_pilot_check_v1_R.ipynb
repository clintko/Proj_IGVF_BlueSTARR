## Check motif summary

```r
### set file directory
txt_fdiry = file.path(FD_RES, "analysis_variant_motif_richard", "motifdelta_pilot")
txt_fname = "variant_closed_gof_bluestarr_flankL35R70_pilot_summary_motif_merged.tsv"
txt_fpath = file.path(txt_fdiry, txt_fname)

### read table
dat = read_tsv(txt_fpath, show_col_types = FALSE)

### assign and show
dat_motif_merge = dat
print(dim(dat))
fun_display_table(head(dat, 3))
```

```r
### set file directory
txt_fdiry = file.path(FD_RES, "analysis_variant_motif_richard", "motifdelta_pilot")

vec_txt_fpath = list.files(txt_fdiry, pattern = "_summary_motif.tsv$", full.names = TRUE)

### import data
dat = purrr::map_dfr(vec_txt_fpath, ~read_tsv(.x, show_col_types = FALSE))

### assign and show
dat_motif_perchunk = dat
print(dim(dat))
fun_display_table(head(dat, 3))
```

```r
### 
dat = dat_motif_perchunk 
dat = dat %>%
    dplyr::group_by(Motif_Name) %>%
    dplyr::summarise(
        Count_Gain = sum(Count_Gain),
        Count_Loss = sum(Count_Loss),
        .groups = "drop"
    ) %>%
    dplyr::arrange(Motif_Name)

### assign and show
dat_motif_remerge = dat
print(dim(dat))
fun_display_table(head(dat, 3))
```

```r
all.equal(
  as.data.frame(dat_motif_remerge),
  as.data.frame(dat_motif_merge),
  check.attributes = FALSE
)
```
