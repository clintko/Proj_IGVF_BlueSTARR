```r
n_var   <- 50
n_motif <- 50

## Top variants by # of gain motifs
vec_variant_gain <- dat_motifdelta_summary_variant %>%
  arrange(desc(Count_Gain)) %>%
  slice_head(n = n_var) %>%
  pull(Variant_ID)

## Top motifs by # of gain variants
vec_motif_gain <- dat_motifdelta_summary_motif %>%
  arrange(desc(Count_Gain)) %>%
  slice_head(n = n_motif) %>%
  pull(Motif_Name)

## Filter to gain events in the chosen subset
dat_gain_sub <- dat_motifdelta_event_all %>%
  filter(
    Event_Type == "Gain",
    Variant_ID %in% vec_variant_gain,
    Motif_Name %in% vec_motif_gain
  )

## Wide matrix: rows = Variant_ID, cols = Motif_Name, value = Delta
mat_gain <- dat_gain_sub %>%
  select(Variant_ID, Motif_Name, Delta) %>%
  tidyr::pivot_wider(
    names_from  = Motif_Name,
    values_from = Delta
  )

## Make Variant_ID the rownames and convert to matrix
mat_gain <- as.data.frame(mat_gain)
rownames(mat_gain) <- mat_gain$Variant_ID
mat_gain$Variant_ID <- NULL
mat_gain <- as.matrix(mat_gain)
```

```r
# top 50 variants by total gain events (from summary_variant)
top_variants_gain <- dat_motifdelta_summary_variant %>% 
  arrange(desc(Count_Gain)) %>% 
  slice_head(n = 1000) %>% 
  pull(Variant_ID)

# top 30 motifs by total gain events (from summary_motif)
top_motifs_gain <- dat_motifdelta_summary_motif %>% 
  arrange(desc(Count_Gain)) %>% 
  slice_head(n = 1000) %>% 
  pull(Motif_Name)

dat_gain_pair_sub <- dat_gain_pair %>% 
  filter(
    Variant_ID %in% top_variants_gain,
    Motif_Name %in% top_motifs_gain
  )

print(dim(dat_gain_pair_sub))
```
