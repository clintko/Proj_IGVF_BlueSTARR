```bash
### set script
FP_EXE=${FD_EXE}/run_benchmark_motifscan.py

### set input files
FP_FASTA=${FD_RES}/analysis_variant_motif_richard/batches_dev/variant_closed_gof_bluestarr_flank35_dev1000_ref.fa
FP_MOTIF=${FD_DATA}/jaspar2024/JASPAR2024_CORE_vertebrates_non-redundant.npz

### init: benchmark parameteres
VEC_TXT_MODE=("base" "serial" "parallel")
#VEC_NUM_LINE=(100 200)
VEC_NUM_LINE=(100 200 400 600 800 1000)

### loop and run benchmark
for TXT_MODE in ${VEC_TXT_MODE[@]}; do
    for NUM_LINE in ${VEC_NUM_LINE[@]}; do
        ### set output file
        FD_OUT=${FD_RES}/analysis_variant_motif_richard/batches_dev
        FP_OUT=${FD_OUT}/benchmark_motifscan_${TXT_MODE}_numseq${NUM_LINE}.csv

        ### set log file
        FP_LOG=${FD_LOG}/run_benchmark_motifscan_${TXT_MODE}_numseq${NUM_LINE}.txt

        ### set resource
        NUM_CPU=10; NUM_MEM=8G; NUM_CORE=10
        
        ### execute
        echo '${FD_LOG}'/"run_benchmark_motifscan_${TXT_MODE}_numseq${NUM_LINE}.txt"
        sbatch \
            -A majoroslab \
            -p igvf \
            --cpus-per-task ${NUM_CPU} \
            --mem    ${NUM_MEM} \
            --output ${FP_LOG}  \
            --chdir  ${FD_EXE}  \
            ${FP_APP} python ${FP_EXE} \
                --txt_fpath_fasta  ${FP_FASTA} \
                --txt_fpath_motif  ${FP_MOTIF} \
                --txt_fpath_output ${FP_OUT} \
                --num_test_seq ${NUM_LINE} \
                --num_core     ${NUM_CORE} \
                --txt_benchmark_mode ${TXT_MODE}
        echo
    done
done
```
