```python
import importlib

packages = {
    "pandas": "pandas",
    "numpy": "numpy",
    "scikit-learn": "sklearn",
    "xgboost": "xgboost",
    "catboost": "catboost",
    "shap": "shap",
    "scipy": "scipy",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
}

print("=== Package check ===")
for label, module_name in packages.items():
    try:
        mod = importlib.import_module(module_name)
        version = getattr(mod, "__version__", "version attribute not found")
        print(f"{label:12s}  OK  version: {version}")
    except ImportError:
        print(f"{label:12s}  MISSING (module name: {module_name})")
```
